```python
import numpy as np
from math import sqrt
```

```python
# Step 1: Sample Data & Mean

data = np.array([1050, 1070, 1065, 1080, 1045, 1075, 1060, 1070])
n = len(data)

sample_mean = np.mean(data)

print("Sample Data:", data)
print("Sample Size (n):", n)
print("Sample Mean (x̄):", sample_mean)
```

```python
# Step 2: Sample Standard Deviation

sample_variance = np.var(data, ddof=1)
sample_std_dev = np.sqrt(sample_variance)

print("\nSample Variance (s²):", sample_variance)
print("Sample Standard Deviation (s):", sample_std_dev)
```

```python
# Step 3: 95% Confidence Interval

t_critical = 2.365  # Given t-value for df = 7

standard_error = sample_std_dev / sqrt(n)
margin_of_error = t_critical * standard_error

ci_lower = sample_mean - margin_of_error
ci_upper = sample_mean + margin_of_error

print("\n95% Confidence Interval:")
print(f"Lower Bound: {ci_lower:.2f}")
print(f"Upper Bound: {ci_upper:.2f}")
```

```python
# Step 4: Interpretation Output

print("\nInterpretation:")
print("We are 95% confident that the true population mean lies within this interval.")
```
