```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 1: Define Transition Matrix

A = np.array([[1.2, 0.1],
              [0.05, 1.1]])

print("Transition Matrix A:")
print(A)
```

```python
# Step 2: Eigenvalue Calculation

eigenvalues, eigenvectors = np.linalg.eig(A)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors (columns correspond to eigenvalues):")
print(eigenvectors)
```

```python
# Identify dominant eigenvalue
dominant_index = np.argmax(eigenvalues)
dominant_eigenvalue = eigenvalues[dominant_index]
dominant_eigenvector = eigenvectors[:, dominant_index]
```

```python
# Normalize dominant eigenvector
dominant_eigenvector = dominant_eigenvector / dominant_eigenvector[0]

print("\nDominant Eigenvalue:")
print(dominant_eigenvalue)

print("\nNormalized Dominant Eigenvector:")
print(dominant_eigenvector)
```

```python
# Step 3: Population Projection

x0 = np.array([1000, 500])  # X0, Y0

time_steps = 20
population = np.zeros((time_steps, 2))
population[0] = x0

for t in range(1, time_steps):
    population[t] = A @ population[t - 1]

print("\nPopulation after 20 time steps:")
print(population[-1])
```

```python
# Step 4: Plot Population Growth

plt.figure()
plt.plot(population[:, 0], label="Population X")
plt.plot(population[:, 1], label="Population Y")
plt.xlabel("Time Steps")
plt.ylabel("Population Size")
plt.title("Population Growth Over Time")
plt.legend()
plt.show()
```

```python
# Step 5: Scenario Analysis

A_modified = np.array([[1.2, 0.1],
                        [0.005, 1.1]])

eigenvalues_mod, _ = np.linalg.eig(A_modified)

print("\nModified Transition Matrix:")
print(A_modified)

print("\nEigenvalues after reducing interaction (β = 0.05):")
print(eigenvalues_mod)

print("\nOriginal Dominant Eigenvalue:", dominant_eigenvalue)
print("New Dominant Eigenvalue:", max(eigenvalues_mod))
```
