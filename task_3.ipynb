```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from math import exp, factorial
```

```python
# Step 1: Define Poisson Model

lam = 5                   
k_values = np.arange(0, 11)
```

```python
# Poisson probability function
def poisson_pmf(k, lam):
    return (exp(-lam) * lam**k) / factorial(k)
```

```python
# Step 2: Calculate Probabilities

probabilities = [poisson_pmf(k, lam) for k in k_values]

poisson_table = pd.DataFrame({
    "Number of Events (k)": k_values,
    "Probability P(X = k)": np.round(probabilities, 4)
})

print("\nPoisson Probability Table (λ = 5):")
print(poisson_table)
```

```python
# Step 3: Expected Value & Variance

expected_value = lam
variance = lam

print("\nExpected Value E[X]:", expected_value)
print("Variance Var[X]:", variance)
```

```python
# Step 4: Scenario Analysis

lam_low = 3
lam_high = 7

prob_low = [poisson_pmf(k, lam_low) for k in k_values]
prob_high = [poisson_pmf(k, lam_high) for k in k_values]
```

```python
# Step 5: Visualization

plt.figure()
plt.plot(k_values, probabilities, marker='o', label="λ = 5 (Baseline)")
plt.plot(k_values, prob_low, marker='o', linestyle='--', label="λ = 3 (Low Immigration)")
plt.plot(k_values, prob_high, marker='o', linestyle=':', label="λ = 7 (High Immigration)")
plt.xlabel("Number of Immigration Events (k)")
plt.ylabel("Probability")
plt.title("Poisson Distribution for Immigration Events")
plt.legend()
plt.show()
```
