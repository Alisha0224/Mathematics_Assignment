```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Define Population Model

P0 = 1000        # Initial population
r = 0.03         # Growth rate (3%)
years = np.arange(0, 21)
```

```python
# Continuous population function
P = P0 * np.exp(r * years)
```

```python
# Step 2: Yearly Population Table

delta_P = np.zeros_like(P)
delta_P[1:] = P[1:] - P[:-1]

population_table = pd.DataFrame({
    "Year (t)": years,
    "Population P(t)": np.round(P, 0).astype(int),
    "Yearly Change ΔP": np.round(delta_P, 0)
})

print("\nPopulation Projection Table:")
print(population_table)
```

```python
# Step 3: Derivative (Growth Rate)

growth_rate = 30 * np.exp(r * years)

print("\nInstantaneous Growth Rates:")
print(f"P'(0)  = {growth_rate[0]:.2f}")
print(f"P'(10) = {growth_rate[10]:.2f}")
print(f"P'(20) = {growth_rate[20]:.2f}")
```

```python
# Step 4: Discrete vs Continuous Model

P_discrete = np.zeros_like(years, dtype=float)
P_discrete[0] = P0

for i in range(1, len(years)):
    P_discrete[i] = P_discrete[i - 1] * (1 + r)
```

```python
# Step 5: Scenario Analysis (r = 0.035)

r_new = 0.035
P_scenario = P0 * np.exp(r_new * years)

print("\nPopulation at Year 20:")
print(f"At r = 0.03  → {P[-1]:.0f}")
print(f"At r = 0.035 → {P_scenario[-1]:.0f}")
```

```python
# Step 6: Cumulative Population (Integral)

cumulative_population = (P0 / r) * (np.exp(r * 20) - 1)

print("\nCumulative Population over 20 years:")
print(f"{cumulative_population:.0f}")
```

```python
# Step 7: Visualization

plt.figure()
plt.plot(years, P, label="Continuous Model")
plt.plot(years, P_discrete, linestyle='--', label="Discrete Model")
plt.plot(years, P_scenario, linestyle=':', label="Scenario (r = 0.035)")
plt.xlabel("Years")
plt.ylabel("Population")
plt.title("Population Growth Models Comparison")
plt.legend()
plt.show()
```
